```python
from senseforge.core import predict

examples = [
    ("I deposited money in the bank.", "bank"),
    ("We sat beside the river bank.", "bank"),
    ("He hit the ball with a bat.", "bat"),
    ("A bat flew out of the cave.", "bat"),
    ("The train arrived at the station.", "train")
]

for sentence, word in examples:
    result = predict(sentence, word, method="dictionary")
    print(f"\nSentence: {sentence}")
    print("Prediction:", result)
```
