```python
from senseforge.core import predict

examples = [
    ("I deposited money in the bank.", "bank"),
    ("We sat beside the river bank.", "bank"),
    ("He hit the ball with a bat.", "bat"),
    ("A bat flew out of the cave.", "bat"),
    ("The train arrived at the station.", "train"),
    ("The coach will train the players.", "train")
]

for sentence, word in examples:
    dictionary = predict(sentence, word, method="dictionary")
    rule = predict(sentence, word, method="rule")

    print("\nSentence:", sentence)
    print("Dictionary:", dictionary["sense_id"], dictionary["hindi"])
    print("Rule-based:", rule["sense_id"], rule["hindi"])
    print("Evidence:", rule.get("evidence"))
```
