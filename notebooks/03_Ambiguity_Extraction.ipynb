## Extract unlabeled target occurrences

```python

import json
import re
from senseforge.core import INVENTORY

source = ART / "clean_parallel.jsonl"
output = ART / "annotation_queue.jsonl"

words = sorted(INVENTORY, key=len, reverse=True)
pattern = re.compile(
    r"(?<!\w)(?:" + "|".join(map(re.escape, words)) + r")(?!\w)",
    re.IGNORECASE
)

count = 0
processed = 0

with source.open("r", encoding="utf-8") as fin, \
     output.open("w", encoding="utf-8") as fout:

    for line in fin:
        row = json.loads(line)
        en = row["en"]
        hi = row["hi"]

        for match in pattern.finditer(en):
            word = match.group().lower()
            record = {
                "id": f"{processed}_{match.start()}",
                "text": en,
                "hindi_sentence": hi,
                "word": word,
                "start": match.start(),
                "end": match.end(),
                "source": row.get("source", "parallel_train"),
                "split": "train",
                "label": None,
                "label_status": "unlabeled"
            }
            fout.write(json.dumps(record, ensure_ascii=False) + "\n")
            count += 1

        processed += 1
        if processed % 100000 == 0:
            print(f"Processed {processed:,} sentences | Candidates {count:,}", flush=True)

print(f"Done: {count:,} candidates saved to {output}")
```
